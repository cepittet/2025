## Collections

```python
#Without Collections
todo_list = {}
if 'ADA' not in todo_list:
    todo_list['ADA'] = list([])
todo_list['ADA'].append('Homework 2')
 
# With Collections
todo_list = collections.defaultdict(list)
todo_list['ADA'].append('Homework 2')
```

```python
# Without Collections
counter = {}
counter['apples'] = 0
counter['oranges'] = 0
counter['apples'] += 1
counter['oranges'] += 1
 
# With Collections
counter = collections.Counter()
counter['apples'] += 1
counter['oranges'] += 1
```

## Strings

```python
#This will convert the integer to string automatically before printing.
num_of_apples = 4

print(num_of_apples)

# You can convert the integers to strings manually, but it's a bad practice.
print("There are " + str(num_of_apples) + " apples on the table.")

# Instead, we recommend you to format strings using %
print("There are %d apples on the table." % (num_of_apples))
```

 **Note:** Use %d for decimals/integers, %f for floats (alternatively %.**x**f to specify a precision of **x**), and %s for strings (or objects with string representations).


## Panda data structure

### Series

```python
# One vector with an index
counts = pd.Series([632, 1638, 569, 115])

# You can get the values
counts.values

# Or the index, if not specified it simply is 0-n
counts.index

# You can also assign the index
counts = pd.Series([632, 1638, 569, 115], index=['First', 'Second', 'Third', 'Fourth'])

# You can then get the value like in a dict
counts['First']

# Another trick
counts[[name.endswith('d') for name in counts.index]]

[name.endswith('d') for name in counts.index] #mask

# You can also apply numpy functions to eat
np.log(counts)
#or
counts.apply(np.log)

# Filter
counts[counts>1000]

# Dict to series
count_dict = { 'First': 632, 
                  'Second': 1638, 
                  'Third': 569,
                  'Fourth': 115 }
pd.Series(count_dict)

```
